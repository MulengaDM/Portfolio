```python
# --- Step 0: Import Necessary Libraries ---
import os
import numpy as np
from skimage.io import imread
from skimage.transform import resize
from zipfile import ZipFile
from google.colab import files
```

```python
# --- Step 1: Upload and Extract the Dataset ---
print("Please upload the 'image classification.zip' file containing 'animal' and 'vehicle' folders.")
uploaded = files.upload()  # Use Google Colab's file upload feature
```

```python
# Extract the ZIP file
zip_file_name = "Image classification.zip"  # Replace this if your file has a different name
with ZipFile(zip_file_name, 'r') as zip_ref:
    zip_ref.extractall("dataset")  # Extract to a 'dataset' folder

print("Dataset extracted successfully!")
```

```python
# --- Step 3: Set Up Paths and Parameters ---
# Update dataset_path to include "Classification"
dataset_path = os.path.join("dataset", "Classification")

categories = ['animal', 'vehicle']  # Folders within "Classification"
IMG_SIZE = 150  # Image size for resizing
```

```python
# Initialize empty lists for data and labels
data = []
labels = []
```

```python
# --- Step 2: Verify the Extracted Directory ---
# Check the directory structure
print("Verifying extracted directory structure...")
for root, dirs, files in os.walk("./dataset"):
    print("Root:", root)
    print("Directories:", dirs)
    print("Files:", files)
    print()
```

```python
# --- Step 4: Load and Preprocess Images ---
print("Processing images...")

for category in categories:
    class_index = categories.index(category)  # Assign labels: 0 for 'animal', 1 for 'vehicle'
    path = os.path.join(dataset_path, category)  # Full path to the category folder

    for img_name in os.listdir(path):
        img_path = os.path.join(path, img_name)

        try:
            # Read and resize the image
            img = imread(img_path)
            img_resized = resize(img, (IMG_SIZE, IMG_SIZE, 3))  # Resize to (150, 150, 3)
            data.append(img_resized)
            labels.append(class_index)
        except Exception as e:
            print(f"Failed to process {img_path}: {e}")

print("Image processing complete!")
```

```python
# --- Step 5: Convert to Numpy Arrays ---
data = np.array(data)
labels = np.array(labels)

print(f"Data shape: {data.shape}")
print(f"Labels shape: {labels.shape}")
```

```python
# --- Step 6: Save the Numpy Arrays ---
np.save("data.npy", data)
np.save("labels.npy", labels)

print("Arrays saved as 'data.npy' and 'labels.npy'.")
```
